```python
df = pd.read_csv("../data/netflix_titles_cleaned.csv")
```

```python
top_ratings = df["rating"].value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_ratings.values, y=top_ratings.index, palette="Pastel1")
plt.title("Top 10 Netflix Ratings")
plt.xlabel("Count")
plt.ylabel("Rating")
plt.show()
```

```python

movies = df[df["type"] == "Movie"].copy()

# Extract the number part, keep only rows with valid numbers
movies["duration_minutes"] = movies["duration"].str.extract(r"(\d+)")

# Drop rows where duration_minutes is NaN
movies = movies.dropna(subset=["duration_minutes"])

# Now convert to int (safe)
movies["duration_minutes"] = movies["duration_minutes"].astype(int)

plt.figure(figsize=(10, 6))
movies["duration_minutes"].hist(bins=30, color="skyblue", edgecolor="black")
plt.title("Distribution of Movie Durations (minutes)")
plt.xlabel("Duration (minutes)")
plt.ylabel("Frequency")
plt.grid(True, alpha=0.3)
plt.show()
```
